````python
%%writefile src/models.py
import datetime
from pydantic import BaseModel, Field


class ExpenseCreate(BaseModel):
    title: str = Field(
        min_length=1,
        max_length=100,
        description="Expense title"
    )

    amount: float = Field(
        gt=0,
        description="Expense amount"
    )

    category: str = Field(
        min_length=1,
        max_length=50,
        description="Expense category"
    )

    date: datetime.date


class Expense(ExpenseCreate):
    id: int
````

````python
%%writefile src/storage.py
import json
import os

BASE_DIR = os.path.dirname(__file__)
FILE_PATH = os.path.join(BASE_DIR, "expenses.json")


def load_data():
    """Load all expenses from JSON file."""
    if not os.path.exists(FILE_PATH):
        return []

    with open(FILE_PATH, "r") as f:
        try:
            return json.load(f)
        except json.JSONDecodeError:
            return []


def save_data(data):
    """Save all expenses to JSON file."""
    with open(FILE_PATH, "w") as f:
        json.dump(data, f, indent=4)


def next_id(data):
    """Generate next expense ID."""
    if not data:
        return 1
    return max(item["id"] for item in data) + 1


def get_expense(expense_id):
    data = load_data()
    for expense in data:
        if expense["id"] == expense_id:
            return expense
    return None


def delete_expense(expense_id):
    data = load_data()

    new_data = [e for e in data if e["id"] != expense_id]

    if len(new_data) == len(data):
        return False

    save_data(new_data)
    return True
````

````python
%%writefile src/main.py
from fastapi import FastAPI, HTTPException, status, Query
from typing import Optional

from src.models import Expense, ExpenseCreate
from src.storage import (
    load_data,
    save_data,
    next_id,
    get_expense,
    delete_expense,
)

app = FastAPI(
    title="💰 Smart Expense Tracker API",
    description="""
A lightweight **RESTful API** built with **FastAPI**
for managing personal expenses.

## Features

- Create Expenses
- View Expenses
- Filter by Category
- Calculate Expense Totals
- Delete Expenses

### Technology

- FastAPI
- Pydantic
- JSON Storage

""",
    version="1.0.0",
    contact={
        "name": "Sandeep Behera",
        "email": "your@email.com",
    },
    license_info={
        "name": "MIT",
    },
)

tags_metadata = [
    {
        "name": "System",
        "description": "Application health endpoints.",
    },
    {
        "name": "Expenses",
        "description": "Create, retrieve and delete expenses.",
    },
    {
        "name": "Analytics",
        "description": "Expense calculations and summaries.",
    },
]


@app.get(
    "/",
    tags=["System"],
    summary="Health Check",
    description="Verify that the Expense Tracker API is running."
)
def home():
    return {"message": "Smart Expense Tracker API is running"}


@app.post(
    "/expenses",
    response_model=Expense,
    status_code=status.HTTP_201_CREATED,
    tags=["Expenses"],
    summary="Create Expense"
)
def add_expense(expense: ExpenseCreate):

    data = load_data()

    new_expense = expense.model_dump()

    new_expense["id"] = next_id(data)
    new_expense["date"] = str(new_expense["date"])

    data.append(new_expense)

    save_data(data)

    return new_expense


@app.get(
    "/expenses",
    tags=["Expenses"],
    summary="List Expenses",
    description="""
Retrieve all expenses.

Optionally filter by category.

Example:

/expenses?category=Food
"""
)
def get_expenses(category: Optional[str] = Query(None)):

    data = load_data()

    if category:
        data = [
            e
            for e in data
            if e["category"].lower() == category.lower()
        ]

    return data
@app.get(
    "/expenses/total",
    tags=["Analytics"],
    summary="Calculate Total Expenses",
    description="""
Calculate the total expenses.

Optionally filter by category.

Example:

/expenses/total?category=Food
"""
)
def total(category: Optional[str] = Query(None)):

    data = load_data()

    if category:
        data = [
            e
            for e in data
            if e["category"].lower() == category.lower()
        ]

    return {
        "count": len(data),
        "total": round(
            sum(e["amount"] for e in data),
            2,
        ),
    }

@app.get(
    "/expenses/{expense_id}",
    tags=["Expenses"],
    summary="Get Expense by ID",
    description="Retrieve a single expense using its unique ID."
)
def get_single_expense(expense_id: int):

    expense = get_expense(expense_id)

    if expense is None:
        raise HTTPException(
            status_code=404,
            detail="Expense not found",
        )

    return expense


@app.delete(
    "/expenses/{expense_id}",
    status_code=status.HTTP_204_NO_CONTENT,
    tags=["Expenses"],
    summary="Delete Expense",
)
def remove_expense(expense_id: int):

    if not delete_expense(expense_id):
        raise HTTPException(
            status_code=404,
            detail="Expense not found"
        )

    return
````

````python
%%writefile tests/test_api.py
import json
import os
import sys

sys.path.insert(0, os.path.abspath("."))

from fastapi.testclient import TestClient
from src.main import app
from src.storage import FILE_PATH

client = TestClient(app)


def setup_function():
    """Reset JSON file before every test."""
    with open(FILE_PATH, "w") as f:
        json.dump([], f)


def test_home():
    response = client.get("/")
    assert response.status_code == 200
    assert "message" in response.json()


def test_add_expense():
    response = client.post(
        "/expenses",
        json={
            "title": "Lunch",
            "amount": 250,
            "category": "Food",
            "date": "2026-08-01"
        },
    )

    assert response.status_code == 201
    data = response.json()

    assert data["id"] == 1
    assert data["title"] == "Lunch"


def test_get_expenses():
    client.post(
        "/expenses",
        json={
            "title": "Lunch",
            "amount": 250,
            "category": "Food",
            "date": "2026-08-01"
        },
    )

    response = client.get("/expenses")

    assert response.status_code == 200
    assert len(response.json()) == 1


def test_filter_category():

    client.post(
        "/expenses",
        json={
            "title": "Lunch",
            "amount": 250,
            "category": "Food",
            "date": "2026-08-01"
        },
    )

    client.post(
        "/expenses",
        json={
            "title": "Bus",
            "amount": 100,
            "category": "Travel",
            "date": "2026-08-01"
        },
    )

    response = client.get("/expenses?category=Food")

    assert response.status_code == 200
    assert len(response.json()) == 1
    assert response.json()[0]["category"] == "Food"


def test_total():

    client.post(
        "/expenses",
        json={
            "title": "Lunch",
            "amount": 250,
            "category": "Food",
            "date": "2026-08-01"
        },
    )

    client.post(
        "/expenses",
        json={
            "title": "Bus",
            "amount": 100,
            "category": "Travel",
            "date": "2026-08-01"
        },
    )

    response = client.get("/expenses/total")

    assert response.status_code == 200
    assert response.json()["total"] == 350


def test_total_food():

    client.post(
        "/expenses",
        json={
            "title": "Lunch",
            "amount": 250,
            "category": "Food",
            "date": "2026-08-01"
        },
    )

    response = client.get("/expenses/total?category=Food")

    assert response.status_code == 200
    assert response.json()["total"] == 250


def test_get_single_expense():

    client.post(
        "/expenses",
        json={
            "title": "Lunch",
            "amount": 250,
            "category": "Food",
            "date": "2026-08-01"
        },
    )

    response = client.get("/expenses/1")

    assert response.status_code == 200
    assert response.json()["title"] == "Lunch"


def test_delete():

    client.post(
        "/expenses",
        json={
            "title": "Lunch",
            "amount": 250,
            "category": "Food",
            "date": "2026-08-01"
        },
    )

    response = client.delete("/expenses/1")

    assert response.status_code == 204


def test_delete_not_found():

    response = client.delete("/expenses/999")

    assert response.status_code == 404


def test_invalid_amount():

    response = client.post(
        "/expenses",
        json={
            "title": "Lunch",
            "amount": -10,
            "category": "Food",
            "date": "2026-08-01"
        },
    )

    assert response.status_code == 422
````
